```python
import os
import re
import pdfplumber
import spacy
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from collections import Counter
from sklearn.feature_extraction.text import TfidfVectorizer

nlp = spacy.load("es_core_news_lg")

# ── STOPWORDS ──────────────────────────────────────
STOPWORDS_JURIDICAS.update({
    "publicado", "julio", "único", "unico", "derogatorio",
    "dejar", "referir", "ámbito", "ambito", "ministro",
    "antiguo", "cesar", "elegir", "máximo", "maximo",
    "periodo", "designación", "designacion", "designar",
    "miércoles", "miercoles"
})

STOPWORDS_EXTRA = STOPWORDS_JURIDICAS.union(nlp.Defaults.stop_words)

# ── PATRONES DE LIMPIEZA ───────────────────────────
PATRONES_RUIDO = [
    r'Artículo\s+\d+[°º]?',                       # Artículo 1°
    r'ARTÍCULO\s+\d+[°º]?',
    r'\b\d+[\.,]?\d*\b',                         # números
    r'ley\s+n[°º]?\s*\d+',                       # ley n° XXXX
    r'decreto\s+legislativo\s+n[°º]?\s*\d+',
    r'decreto\s+supremo\s+n[°º]?\s*\d+',
    r'resolución\s+\w+\s+n[°º]?\s*\d+',
    r'\b[a-z]{1,2}\b',                           # tokens de 1-2 letras
    r'http\S+|www\.\S+',                         # urls
]

# ── FUNCIONES ──────────────────────────────────────
def extraer_documento(meta):
    texto = ""
    with pdfplumber.open(meta["archivo"]) as pdf:
        for p in pdf.pages:
            if meta.get("dos_columnas"):
                w, h = p.width, p.height
                texto += (p.crop((0, 0, w/2, h)).extract_text() or "") + " "
                texto += (p.crop((w/2, 0, w, h)).extract_text() or "") + " "
            else:
                texto += (p.extract_text() or "") + " "

    ini = meta.get("marcador_inicio")
    fin = meta.get("marcador_fin")

    if ini and fin:
        i0 = texto.find(ini)
        i1 = texto.find(fin)
        if i0 == -1:
            print("  ⚠️ Inicio no encontrado")
            return ""
        if i1 == -1:
            print("  ⚠️ Fin no encontrado")
            return ""
        return texto[i0:i1 + len(fin)]

    return texto


def limpiar(texto):
    texto = texto.lower()

    # normalización básica
    texto = texto.replace("\xad", " ")  # guion de corte
    texto = texto.replace("“", '"').replace("”", '"')
    texto = texto.replace("’", "'").replace("‘", "'")

    # eliminar patrones de ruido
    for patron in PATRONES_RUIDO:
        texto = re.sub(patron, " ", texto, flags=re.IGNORECASE)

    # dejar solo letras y espacios, conservando tildes/ñ
    texto = re.sub(r'[^a-záéíóúüñ\s]', ' ', texto)

    # espacios múltiples
    texto = re.sub(r'\s+', ' ', texto).strip()

    return texto


def tokenizar(texto):
    doc = nlp(texto)
    tokens = []

    for t in doc:
        lemma = t.lemma_.strip().lower()

        if (
            t.is_stop
            or t.is_punct
            or t.is_space
            or t.like_num
            or len(lemma) <= 3
            or lemma in STOPWORDS_EXTRA
        ):
            continue

        # filtrar basura residual
        if not re.fullmatch(r'[a-záéíóúüñ]+', lemma):
            continue

        tokens.append(lemma)

    return tokens


def reconstruir_texto(tokens):
    return " ".join(tokens)


def calcular_mi(tokens, focales, ventana=5, min_freq=3):
    N = len(tokens)
    freq = Counter(tokens)
    colocados, freq_focal = [], 0
    encontrados = []

    for f in focales:
        if freq.get(f, 0) == 0:
            continue
        encontrados.append(f)
        freq_focal += freq[f]

        for i, t in enumerate(tokens):
            if t == f:
                w = tokens[max(0, i-ventana):i] + tokens[i+1:i+ventana+1]
                colocados.extend(w)

    if not encontrados:
        return [], []

    fc = Counter(colocados)
    res = {}

    for p, fconj in fc.items():
        if freq[p] < min_freq or p in focales:
            continue

        pconj = fconj / N
        pfoc = freq_focal / N
        ppal = freq[p] / N

        if pfoc * ppal > 0:
            res[p] = round(np.log2(pconj / (pfoc * ppal)), 3)

    return sorted(res.items(), key=lambda x: x[1], reverse=True), encontrados


# ── EXTRACCIÓN Y PREPROCESAMIENTO ──────────────────
nombres_ordenados = sorted(CORPUS, key=lambda x: CORPUS[x]["momento"])
textos_limpios, tokens_por_doc, textos_para_tfidf = {}, {}, {}

print("EXTRACCIÓN Y PREPROCESAMIENTO\n" + "="*50)

for nombre in nombres_ordenados:
    meta = CORPUS[nombre]
    print(f"\n{nombre} — {meta['label']}")

    raw = extraer_documento(meta)
    if not raw:
        continue

    limpio = limpiar(raw)
    tokens = tokenizar(limpio)
    texto_tfidf = reconstruir_texto(tokens)

    textos_limpios[nombre] = limpio
    tokens_por_doc[nombre] = tokens
    textos_para_tfidf[nombre] = texto_tfidf

    freq = Counter(tokens)

    print(f"  ✓ Tokens limpios: {len(tokens)}")
    print(f"  ✓ Vocabulario: {len(set(tokens))}")

    for t in meta["terminos_equivalentes"]["unidad_institucional"]:
        if freq.get(t, 0) > 0:
            print(f"      '{t}': {freq[t]}")

    print("  Top 10 preliminar:")
    for tok, f in freq.most_common(10):
        print(f"      {tok:<20} {f}")


# ── TF-IDF: PRIMERA PASADA (UNIGRAMAS) ─────────────
print("\nTF-IDF (UNIGRAMAS)\n" + "="*50)

textos_ord = [textos_para_tfidf[n] for n in nombres_ordenados]

vec_uni = TfidfVectorizer(
    max_features=300,
    min_df=1,
    max_df=0.85,
    ngram_range=(1, 1)
)

matriz_uni = vec_uni.fit_transform(textos_ord)
df_tfidf_uni = pd.DataFrame(
    matriz_uni.toarray(),
    index=nombres_ordenados,
    columns=vec_uni.get_feature_names_out()
)

resultados_tfidf_uni = {}

for nombre in nombres_ordenados:
    top20 = df_tfidf_uni.loc[nombre].nlargest(20)
    resultados_tfidf_uni[nombre] = top20
    print(f"\n{nombre} — {CORPUS[nombre]['label']}")
    for t, v in top20.items():
        print(f"  {t:<35} {v:.4f}")


# ── VISUALIZACIÓN TF-IDF UNIGRAMAS ─────────────────
fig, axes = plt.subplots(2, 2, figsize=(16, 12))
colores = ["#2196F3", "#FF9800", "#4CAF50", "#F44336"]

for idx, nombre in enumerate(nombres_ordenados):
    ax = axes.flatten()[idx]
    top15 = resultados_tfidf_uni[nombre].head(15)

    ax.barh(range(len(top15)), top15.values, color=colores[idx], alpha=0.8)
    ax.set_yticks(range(len(top15)))
    ax.set_yticklabels(top15.index, fontsize=9)
    ax.invert_yaxis()
    ax.set_title(
        f"Momento {CORPUS[nombre]['momento']}: {CORPUS[nombre]['label']}",
        fontsize=11, fontweight="bold"
    )
    ax.grid(axis="x", alpha=0.3, linestyle="--")

plt.suptitle("Términos distintivos por momento regulatorio (unigramas)",
             fontsize=14, fontweight="bold")
plt.tight_layout()
plt.savefig(os.path.join(RUTA_OUTPUTS, "tfidf_comparativo_unigramas.png"),
            dpi=150, bbox_inches="tight")
plt.show()


# ── TF-IDF: SEGUNDA PASADA (UNIGRAMAS + BIGRAMAS) ──
print("\nTF-IDF (UNIGRAMAS + BIGRAMAS)\n" + "="*50)

vec_bi = TfidfVectorizer(
    max_features=400,
    min_df=1,
    max_df=0.85,
    ngram_range=(1, 2)
)

matriz_bi = vec_bi.fit_transform(textos_ord)
df_tfidf_bi = pd.DataFrame(
    matriz_bi.toarray(),
    index=nombres_ordenados,
    columns=vec_bi.get_feature_names_out()
)

resultados_tfidf_bi = {}

for nombre in nombres_ordenados:
    top20 = df_tfidf_bi.loc[nombre].nlargest(20)
    resultados_tfidf_bi[nombre] = top20
    print(f"\n{nombre} — {CORPUS[nombre]['label']}")
    for t, v in top20.items():
        print(f"  {t:<35} {v:.4f}")
```

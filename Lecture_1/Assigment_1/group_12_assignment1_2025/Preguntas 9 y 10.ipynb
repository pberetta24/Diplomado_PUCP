```python
def resumen_peliculas(peliculas):
    resumenes = []
    for titulo, detalles in peliculas.items():
        oracion = f"La película {titulo} pertenece al género {detalles['genre']} y fue estrenada en {detalles['year']}."
        resumenes.append(oracion)
    return resumenes

# Ejemplo de uso:
peliculas = {
    "Inception": {"year": 2010, "genre": "Ciencia Ficción"},
    "Parasite": {"year": 2019, "genre": "Drama"},
    "Whiplash": {"year": 2014, "genre": "Drama"}
}

print(resumen_peliculas(peliculas))
```
